```python
# Don't let the business logic worry about 
# which concrete object to create. 
# Let a factory handle object creation.

class EmailNotification:
    def send(self, message):
        print(f"Email: {message}")

class SMSNotification:
    def send(self, message):
        print(f"SMS: {message}")


# create factory
class NotificationFactory:
    @staticmethod
    def create(notification_type) -> EmailNotification | SMSNotification:
        if notification_type == "email":
            return EmailNotification()
        if notification_type == "sms":
            return SMSNotification()
        raise ValueError("Unknown notification type")


notification= NotificationFactory.create("email")
notification.send("Hello!")
```

```python
from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send(self, message):
        pass

class EmailNotification(Notification):
    def send(self, message):
        print(f"Email: {message}")

class SMSNotification(Notification):
    def send(self, message):
        print(f"SMS: {message}")

# Here the base class defines the factory method:
class NotificationService(ABC):
    @abstractmethod
    def create_notification(self):
        pass
    
    def notify(self, message):
        notification: None = self.create_notification()
        notification.send(message)


class EmailNotificationService(NotificationService):
    def create_notification(self):
        return EmailNotification()

class SMSNotificationService(NotificationService):
    def create_notification(self):
        return SMSNotification()



service = EmailNotificationService()
service.notify("Hello!")
```

```python
from abc import ABC, abstractmethod


class LLM(ABC):
    @abstractmethod
    def generate(self, prompt):
        pass

class Embedding(ABC):
    @abstractmethod
    def embed(self, text):
        pass

class OpenAILLM(LLM):
    def generate(self, prompt):
        return "OpenAI response"

class OpenAIEmbedding(Embedding):
    def embed(self, text):
        return [0.1, 0.2, 0.3]

class GeminiLLM(LLM):
    def generate(self, prompt):
        return "Gemini response"

class GoogleEmbedding(Embedding):
    def embed(self, text):
        return [0.4, 0.5, 0.6]
```

```python
class AIProviderFactory(ABC):

    @abstractmethod
    def create_llm(self) -> LLM:
        pass

    @abstractmethod
    def create_embedding(self) -> Embedding:
        pass

class OpenAIFactory(AIProviderFactory):

    def create_llm(self):
        return OpenAILLM()

    def create_embedding(self):
        return OpenAIEmbedding()

class GoogleFactory(AIProviderFactory):

    def create_llm(self):
        return GeminiLLM()

    def create_embedding(self):
        return GoogleEmbedding()


factory = GoogleFactory()

llm = factory.create_llm()
embedding = factory.create_embedding()
```

## builder design pattern

```python
class Computer:

    def __init__(
        self,
        cpu,
        ram,
        storage,
        gpu=None,
        wifi=False
    ):
        self.cpu = cpu
        self.ram = ram
        self.storage = storage
        self.gpu = gpu
        self.wifi = wifi


computer = Computer(
    cpu="Intel i7",
    ram="32GB",
    storage="1TB SSD",
    gpu="RTX 5090",
    wifi=True
)
```

```python
class ComputerBuilder:

    def __init__(self) :
        self.cpu = None
        self.ram = None
        self.storage = None
        self.gpu = None
        self.wifi = False

    def cpu(self, cpu):
        self.cpu = cpu
        return self

    def ram(self, ram):
        self.ram = ram
        return self

    def storage(self, storage):
        self.storage = storage
        return self

    def gpu(self, gpu):
        self.gpu = gpu
        return self

    def enable_wifi(self):
        self.wifi = True
        return self

    def build(self):
        return Computer(
            cpu=self.cpu,
            ram=self.ram,
            storage=self.storage,
            gpu=self.gpu,
            wifi=self.wifi
        )

computer = (
    ComputerBuilder()
    .cpu("Intel i7")
    .ram("32GB")
    .storage("1TB SSD")
    .gpu("RTX 5090")
    .enable_wifi()
    .build()
)
```

In Python Object-Oriented Programming, __new__ is the special method responsible for actually creating a new instance of a class.
While most developers think __init__ is the constructor, __init__ is actually just an initializer that sets up attributes on an object that already exists. It is __new__ that acts as the true creator method, running before __init__ to allocate memory for the object.
## The Lifecycle: __new__ vs __init__
When you write obj = MyClass(), Python completes two distinct steps behind the scenes:

   1. __new__(cls, ...): Runs first. It takes the class itself as an argument, builds the raw object in memory, and must return that object.
   2. __init__(self, ...): Runs second. It takes the newly created object (self) returned by __new__ and customizes it by assigning initial variables.

Here is a direct comparison of their differences:

| Feature | __new__ | __init__ |
|---|---|---|
| Role | Object Creator (Allocates memory) | Object Initializer (Assigns data) |
| Execution Order | Runs first | Runs second |
| First Argument | cls (The class itself) | self (The instance just created) |
| Return Value | Must return a new object instance | Must NOT return anything (returns None) |

------------------------------
## Code Example: Overriding __new__
You rarely need to override __new__ unless you want to intercept and alter how an object is built. A classic use case is implementing the Singleton Pattern, ensuring a class can only ever create one single instance.


## When should you actually use __new__?
You should leave __new__ alone and stick to __init__ for 99% of your coding tasks. However, you will need __new__ if you are:

* Subclassing immutable types: If you want to modify a built-in immutable class like str, int, or tuple, you cannot change them in __init__ because they are already locked. You must alter them during creation in __new__.
* Design Patterns: Implementing strict architectural boundaries like Singletons, or caching objects to save memory.
* Metaprogramming: Controlling complex framework-level class generation.

```python
class DatabaseConnection:
    _instance = None  # Class variable to store our single instance

    def __new__(cls, *args, **kwargs):
        print("1. __new__ called: Allocating memory for the object.")
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, db_name):
        print(f"2. __init__ called: Customizing the instance with name: {db_name}\n")
        self.db_name = db_name


# Attempt to create two separate connections
conn1 = DatabaseConnection("Production_DB")
conn2 = DatabaseConnection("Testing_DB")

# Proof that both variables point to the exact same object in memory
print(f"Is conn1 the same instance as conn2? {conn1 is conn2}")  
# Output: True

print(f"conn1 name: {conn1.db_name}")  # Output: Testing_DB
print(f"conn2 name: {conn2.db_name}")  # Output: Testing_DB
```

## prototype designe pattern

```python
class User:

    def __init__(self, name, role):
        self.name = name
        self.role = role

user1 = User("Manish", "Engineer")
user2 = User("Rahul", "Engineer")
```

```python
# We created user2 from the existing user1 object.
# Instead of creating a new object from scratch, create a new object
#  by copying an existing object (prototype).
# ototype becomes useful when creating an object is expensive or complicated.

import copy
user2 = copy.copy(user1)
user2.name = "Rahul"
```

```python
print(user1.name)
print(user1.role)

print(user2.name)
print(user2.role)
```

```python
from abc import ABC, abstractmethod
import copy


class Prototype(ABC):

    @abstractmethod
    def clone(self):
        pass

class Agent(Prototype):

    def __init__(self, name, model):
        self.name = name
        self.model = model

    def clone(self):
        return copy.deepcopy(self)



research_agent = Agent(
    "research_agent",
    "gemini"
)

market_agent = research_agent.clone()

market_agent.name = "market_agent"
```
