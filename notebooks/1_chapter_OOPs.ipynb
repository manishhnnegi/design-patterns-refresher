## encapsulastion

```python
#Encapsulation hides the internal state of the object and allows controlled access via methods.
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance  # Private attribute

    def deposit(self, amount):
        self._balance += amount

    def withdraw(self, amount):
        if amount <= self._balance:
            self._balance -= amount
        else:
            print("Insufficient funds.")

    def get_balance(self):
        return self._balance

account = BankAccount("Kunnu", 1000)
account.deposit(500)
account.withdraw(300)
print(f"Balance: {account.get_balance()}")
```

```python
#Inheritance allows a class to derive properties and methods from another class.

class Animal:
    def speak(self):
        return "I am an animal."

class Dog(Animal):  # Dog inherits from Animal
    def speak(self):
        return "Woof! Woof!"

dog = Dog()
print(dog.speak())
```

```python
#Polymorphism allows objects of different classes to be treated as objects of a

class Bird:
    def sound(self):
        return "Some generic bird sound."

class Sparrow(Bird):
    def sound(self):
        return "Chirp Chirp."

class Crow(Bird):
    def sound(self):
        return "Caw Caw."

birds = [Sparrow(), Crow()]

for bird in birds:
    print(bird.sound())
```

```python
# Abstraction hides implementation details and exposes only functionality.
# Use the abc module to create abstract classes.

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

rect = Rectangle(4, 5)
print(f"Area of rectangle: {rect.area()}")
```

```python
class Shape(ABC):

    @abstractmethod
    def area(self):
     pass

    # no self is used 
    # can not access by the object of the class
    # its a utility method inside a class

    @staticmethod
    def describe_shape():
     return "This is a shape with an area and a perimeter."


    # This decorator is used to define a class method that operates on the class itself (not an
    # instance)
    @classmethod
    def create_rectangle(cls, width, height):
        return Rectangle(width, height)


# @staticmethod does not have access to the class or instance.
# @classmethod has access to the class
```

```python
rect = Shape.create_rectangle(4, 5)
print(rect.area())  # Output: 20
```

```python
class Rectangle:
    def __init__(self, width, height):
        self._width = width
        self._height = height

    @property
    def area(self):
        return self._width * self._height  # Computed property

# Usage
rect = Rectangle(4, 5)
print(rect.area)  # Output: 2
```

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int
# Usage

p1 = Point(1, 2)
print(p1)  # Output: Point(x=1, y=2)
```

```python
class MyClass:
    def __init__(self):
        self.__private_var = 42
    def __private_method(self):
        return "This is a private method."
    def access_private_method(self):
        return self.__private_method()
# Usage

obj = MyClass()
print(obj.access_private_method())  # Output: This is a private method
```

```python
# Allows an instance of a class to be called as if it were a function
class Counter:
    def __init__(self, start=0):
        self.count = start
    def __call__(self):
        self.count += 1
        return self.count
# Creating an instance of Counter
counter = Counter()
print(counter())  # Output: 1
print(counter())  # Output: 
```
